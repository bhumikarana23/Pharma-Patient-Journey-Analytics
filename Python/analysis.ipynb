```python
from scipy.stats import ttest_ind

switched = df[df["Switched_Therapy"] == 1]["HbA1c"]
not_switched = df[df["Switched_Therapy"] == 0]["HbA1c"]

t_stat, p_value = ttest_ind(switched, not_switched)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("The difference in HbA1c between switched and non-switched patients is statistically significant.")
else:
    print("The difference in HbA1c between switched and non-switched patients is not statistically significant.")
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["Switched_Therapy"])
y = df["Switched_Therapy"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
features = [
    "Age",
    "Sex",
    "Disease_Duration_Years",
    "Disease_Severity",
    "Previous_Therapy",
    "Current_Therapy",
    "Treatment_Duration_Days",
    "Annual_Visits",
    "HbA1c",
    "BMI",
    "Medication_Adherence_Pct",
    "Adverse_Event",
    "Treatment_Cost_Index"
]

X = df[features]
y = df["Switched_Therapy"]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nFeatures:")
print(X.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "Sex",
    "Disease_Severity",
    "Previous_Therapy",
    "Current_Therapy",
    "Adverse_Event"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)

X_encoded = preprocessor.fit_transform(X)

print("Original features:", X.shape)
print("Encoded features:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-Score:", f1_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Switch", "Switch"]
)

disp.plot()
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

model_balanced.fit(X_train, y_train)

y_pred_balanced = model_balanced.predict(X_test)

print("Balanced Logistic Regression trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy:", accuracy_score(y_test, y_pred_balanced))
print("Precision:", precision_score(y_test, y_pred_balanced))
print("Recall:", recall_score(y_test, y_pred_balanced))
print("F1-Score:", f1_score(y_test, y_pred_balanced))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1-Score:", f1_score(y_test, y_pred_rf))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

y_prob = model_balanced.predict_proba(X_test)[:, 1]

auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", auc)
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Balanced Logistic Regression")
plt.show()
```

```python
# Get feature names after encoding
feature_names = preprocessor.get_feature_names_out()

# Get model coefficients
coefficients = model_balanced.coef_[0]

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

# Absolute coefficient = strength of association
feature_importance["Importance"] = feature_importance["Coefficient"].abs()

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

print(feature_importance.head(15))
```

```python
feature_importance["Direction"] = feature_importance["Coefficient"].apply(
    lambda x: "Higher switching likelihood" if x > 0 else "Lower switching likelihood"
)

print(feature_importance.head(15)[
    ["Feature", "Coefficient", "Direction"]
])
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

segmentation_features = [
    "Age",
    "Disease_Duration_Years",
    "Treatment_Duration_Days",
    "Annual_Visits",
    "HbA1c",
    "BMI",
    "Medication_Adherence_Pct",
    "Treatment_Cost_Index"
]

X_seg = df[segmentation_features]

scaler = StandardScaler()
X_seg_scaled = scaler.fit_transform(X_seg)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

df["Patient_Segment"] = kmeans.fit_predict(X_seg_scaled)

print(df["Patient_Segment"].value_counts().sort_index())
```

```python
segment_profile = df.groupby("Patient_Segment")[segmentation_features].mean().round(2)

print(segment_profile)
```
